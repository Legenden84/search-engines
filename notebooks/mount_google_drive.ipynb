### Injest PyTerrier Indexes into Colab CPU

```python
src = Path("/content/drive/MyDrive/se_project/indexes")
dst = Path("/content/indexes")

if not src.exists():
    raise FileNotFoundError(f"Saved indexes not found at: {src}")

if dst.exists():
    print("Removed existing PyTerrier indexes")
    shutil.rmtree(dst)

shutil.copytree(src, dst)

print(f"Copied saved indexes from {src} to {dst}")
```

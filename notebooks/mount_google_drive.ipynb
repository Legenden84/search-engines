# Imports

```python
import shutil
from pathlib import Path

# Folder where the notebook is running
PROJECT_DIR = Path.cwd() / "msc_datalogi" / "2_semester" / "se"

# Persistent local cache folder on your Jupyter server
CACHE_DIR = Path.home() / "msc_datalogi" / "2_semester" / "se_project_cache"

print(f"Project directory: {PROJECT_DIR}")
print(f"Cache directory:   {CACHE_DIR}")
```

# Injest PyTerrier Indexes into Colab CPU

```python
folders = ["indexes", "results"]

for folder in folders:
    src = CACHE_DIR / folder
    dst = PROJECT_DIR / folder

    if not src.exists():
        print(f"Skipping. Saved folder not found: {src}")
        continue

    if dst.exists():
        print(f"Removing existing {folder} folder at {dst}")
        shutil.rmtree(dst)

    shutil.copytree(src, dst)

    print(f"Copied saved {folder}: {src} -> {dst}")
```

# Copy Colab CPU files to Google Drive

```python
folders = ["indexes", "results"]

for folder in folders:
    src = Path(f"/content/{folder}")
    dst = Path(f"/content/drive/MyDrive/se_project/{folder}")

    if not src.exists():
        print(f"Skipping. Source folder not found: {src}")
        continue

    shutil.copytree(src, dst, dirs_exist_ok=True)

    print(f"Copied {folder}: {src} -> {dst}")
```
